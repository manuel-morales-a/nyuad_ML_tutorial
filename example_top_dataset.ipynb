In this example we will work with the public [Top Quark Tagging Reference Dataset](https://zenodo.org/records/2603256), which provides the following description:

A set of MC simulated training/testing events for the evaluation of top quark tagging architectures.

In total 1.2M training events, 400k validation events and 400k test events. Use “train” for training, “val” for validation during the training and “test” for final testing and reporting results.

Description

* 14 TeV, hadronic tops for signal, qcd diets background, Delphes ATLAS detector card with Pythia8

* No MPI/pile-up included

* Clustering of  particle-flow entries (produced by Delphes E-flow) into anti-kT 0.8 jets in the pT range [550,650] GeV

* All top jets are matched to a parton-level top within ∆R = 0.8, and to all top decay partons within 0.8

* Jets are required to have |eta| < 2

* The leading 200 jet constituent four-momenta are stored, with zero-padding for jets with fewer than 200

* Constituents are sorted by pT, with the highest pT one first

* The truth top four-momentum is stored as truth_px etc.

* A flag (1 for top, 0 for QCD) is kept for each jet. It is called is_signal_new

* The variable "ttv" (= test/train/validation) is kept for each jet. It indicates to which dataset the jet belongs. It is redundant as the different sets are already distributed as different files.

Let's compute the average $p_T$ (it's not the best observable but it's faster to compute). To do this, we need to move from (E,px,py,pz) to (pT,eta,phi,mass).

```python
def Epxpypz_to_pTetaphimass(vec):
    E, px, py, pz = vec[0], vec[1], vec[2], vec[3]
    pT = np.sqrt(px**2+py**2)
    phi = np.arctan2(py,px)
    mass = np.sqrt(E**2-px**2-py**2-pz**2) if E**2-px**2-py**2-pz**2 > 0 else 0
    eta = np.arctanh(pz/np.sqrt(px**2+py**2+pz**2))
    return np.array([pT,eta,phi,mass])
```

```python
# pT of every constituent, then average over the real (non-padded) ones using the mask
constituent_pT = np.sqrt(train_features[:,1:801:4]**2+train_features[:,2:801:4]**2)
av_pT = np.sum(constituent_pT*train_mask,axis=1)/multiplicity
```

# A home-made DeepSets classifier: [ParticleFlow Network](http://arxiv.org/abs/1810.05165)

We will do a ParticleFlow network but in 4-momenta space, without using the ($p_{T}$,$\eta$,$\phi$) representation. This will probably cost some performance, but it takes a bit longer to preprocess 1M events. You probably want to do that (and maybe try a CNN there as well!)

We'll define a custom Dataset and a custom Neural Network class in Pytorch.

Our dataset should have features, label and mask to indicate which particles are zero-padded

```python
class CustomDataset(torch.utils.data.Dataset):
    """
    Joins the x and y into a dataset, so that it can be used by the pythorch syntax.
    """

    def __init__(self, x, y,m):
        self.x = torch.tensor(x).to(torch.float)
        self.y = torch.tensor(y).to(torch.float)
        self.m = torch.tensor(m).to(torch.float)

    def __len__(self):
        return len(self.x)

    def __getitem__(self, idx):
        if torch.is_tensor(idx):
            idx = idx.tolist()
        sample = [self.x[idx], self.y[idx],self.m[idx]]
        return sample
```

Our model consists of two neural networks, one evaluated per particle and one evaluated over all particles in an event

```python
from torch import nn
from torch.nn.modules import Module
from torch.utils.data import DataLoader, BatchSampler, RandomSampler, SequentialSampler

if torch.cuda.is_available():
    device = torch.device("cuda")  # NVIDIA GPU, e.g. on Colab
elif torch.backends.mps.is_available():
    device = torch.device("mps")   # Apple Silicon GPU
else:
    device = torch.device("cpu")
print("Using device:", device)
```

```python

# Define model
class PseudoParticleFlow(nn.Module):
    def __init__(self,dim_input=4,inner_dim=64,layers_Phi=[(64, nn.ReLU()),(64, nn.ReLU()),(64, nn.ReLU())],layers_F=[(64, nn.ReLU()),(64, nn.ReLU()),(64, nn.ReLU())]): # example of layers_data=[(layer1, nn.ReLU()), (layer2, nn.ReLU()), (output_size, nn.Sigmoid())]
        super(PseudoParticleFlow, self).__init__()
        self.flatten = nn.Flatten()
        self.Phi = nn.ModuleList()
        self.F = nn.ModuleList()
        self.input_size = dim_input  # Can be useful later ...
        self.inner_dim = inner_dim  # Can be useful later ...
        input_size = dim_input
        for size, activation in layers_Phi:
            self.Phi.append(nn.Linear(input_size, size))
            input_size = size  # For the next layer
            if activation is not None:
                assert isinstance(activation, Module), \
                    "Each tuples should contain a size (int) and a torch.nn.modules.Module."
                self.Phi.append(activation)
        self.Phi.append(nn.Linear(input_size, inner_dim))

        input_size = inner_dim
        for size, activation in layers_F:
            self.F.append(nn.Linear(input_size, size))
            input_size = size  # For the next layer
            if activation is not None:
                assert isinstance(activation, Module), \
                    "Each tuples should contain a size (int) and a torch.nn.modules.Module."
                self.F.append(activation)
        self.F.append(nn.Linear(input_size, 1))
        self.F.append(nn.Sigmoid())

    def forward(self, x,m):
        output = x
        for layer in self.Phi:
            output = layer(output)
            #print(output.shape)
        ### this uses the mask to remove zero-padded particles when performing the averaging
        output = torch.einsum('ijk,ij->ik',output,m)/(torch.einsum('ik,ik->i',m,m).unsqueeze(1))
        #output = torch.stack((torch.masked.masked_tensor(output[:,:,0], m),torch.masked.masked_tensor(output[:,:,0], m)),dim=-1)
        #print(output.shape)
        #output = torch.mean(output,1)
        #print(output.shape)
        for layer in self.F:
            output = layer(output)
            #print(output.shape)
        return output

    def latent(self, x,m):
        output = x
        for layer in self.Phi:
            output = layer(output)
            #print(output.shape)
        ### this uses the mask to remove zero-padded particles when performing the averaging
        output = torch.einsum('ijk,ij->ik',output,m)/(torch.einsum('ik,ik->i',m,m).unsqueeze(1))
        return output

    def proba_from_latent(self, ell):
        output = ell
        for layer in self.F:
            output = layer(output)
            #print(output.shape)
        return output

    def loss_function(self,y,t):
        #loss_fn = nn.MSELoss(reduction='mean')
        loss_fn = nn.BCELoss()

        return loss_fn(y,t)

    def reset_weights(self):
        for m in self.Phi:
            if isinstance(m, nn.Conv2d) or isinstance(m, nn.Linear):
                m.reset_parameters()
        for m in self.F:
            if isinstance(m, nn.Conv2d) or isinstance(m, nn.Linear):
                m.reset_parameters()
```

We can define training, testing and predicting functions

```python
def make_loader(dataset, batch_size, shuffle):
    """Batches of `batch_size` jets. Sampling whole batches of indices at once (instead of
    one jet at a time) is much faster, since our dataset can slice many jets in one go."""
    sampler = RandomSampler(dataset) if shuffle else SequentialSampler(dataset)
    return DataLoader(dataset, sampler=BatchSampler(sampler, batch_size, drop_last=False), batch_size=None)

def train(model,optimizer,train_dataset,batch_size=1024):
    model.train()
    train_dataset_batched = make_loader(train_dataset,batch_size,shuffle=True)
    total_loss = 0
    for batch, (X, y,M) in enumerate(train_dataset_batched):
        X, y, M  = X.to(device), y.to(device), M.to(device)
        #print(torch.mean(y))
        optimizer.zero_grad()  # Clear gradients.
        pred = model(X,M)[:,0]  # Forward pass.
        #print(pred.shape,y.shape)
        loss = model.loss_function(pred,y)  # Loss computation.
        loss.backward()  # Backward pass.
        optimizer.step()  # Update model parameters.
        total_loss += loss.item()

    return total_loss / len(train_dataset_batched)

@torch.no_grad()
def test(model, test_dataset,batch_size=1024):
    model.eval()
    test_dataset_batched = make_loader(test_dataset,batch_size,shuffle=False)
    total_loss = 0
    for batch, (X, y,M) in enumerate(test_dataset_batched):
        X, y, M  = X.to(device), y.to(device), M.to(device)
        pred = model(X,M)[:,0]  # Forward pass.
        loss = model.loss_function(pred,y)  # Loss computation.
        total_loss += loss.item()

    return total_loss / len(test_dataset_batched)

@torch.no_grad()
def predict_proba(model,test_dataset,batch_size=1024):
    model.eval()
    test_dataset_batched = make_loader(test_dataset,batch_size,shuffle=False)
    truths = np.zeros(len(test_dataset))
    preds = np.zeros(len(test_dataset))
    initial_index = 0
    final_index = 0
    for batch, (X, y,M) in enumerate(test_dataset_batched):
        final_index+=len(y)
        truths[initial_index:final_index]=y.detach().cpu().numpy()
        X, y, M  = X.to(device), y.to(device), M.to(device)
        pred = model(X,M)[:,0]  # Forward pass.
        preds[initial_index:final_index] = pred.detach().cpu().numpy()
        initial_index+=len(y)
    return truths, preds

@torch.no_grad()
def predict(model,test_dataset,batch_size=1024):
    model.eval()
    test_dataset_batched = make_loader(test_dataset,batch_size,shuffle=False)
    truths = np.zeros(len(test_dataset))
    preds = np.zeros(len(test_dataset))
    initial_index = 0
    final_index = 0
    for batch, (X, y,M) in enumerate(test_dataset_batched):
        final_index+=len(y)
        truths[initial_index:final_index]=y.detach().cpu().numpy()
        X, y, M  = X.to(device), y.to(device), M.to(device)
        pred = model(X,M)[:,0]  # Forward pass.
        preds[initial_index:final_index] = pred.detach().cpu().numpy()
        preds[initial_index:final_index] = np.where(preds[initial_index:final_index]>0.5,1,0)
        initial_index+=len(y)
    return truths, preds

@torch.no_grad()
def latent(model,test_dataset,batch_size=1024):
    model.eval()
    test_dataset_batched = make_loader(test_dataset,batch_size,shuffle=False)
    latent_variables = np.zeros((len(test_dataset),model.inner_dim))
    initial_index = 0
    final_index = 0
    for batch, (X, y,M) in enumerate(test_dataset_batched):
        final_index+=len(y)
        X, y, M  = X.to(device), y.to(device), M.to(device)
        pred = model.latent(X,M)  # Forward pass.
        latent_variables[initial_index:final_index] = pred.detach().cpu().numpy()
        initial_index+=len(y)
    return latent_variables
```

## Training the network: from scratch, or load a pre-trained one

Training the network for 30 epochs takes a few minutes on a GPU (and much longer on a CPU), so **by default we load a model that we trained beforehand with exactly this notebook** (same 300k training jets, same settings), together with its training history. If you want to train it yourself, set `TRAIN_FROM_SCRATCH = True` below and run the next cells (use a GPU runtime!).

```python
TRAIN_FROM_SCRATCH = False  # set to True to train the network yourself

PRETRAINED_URL = "https://raw.githubusercontent.com/manuel-morales-a/nyuad_ML_tutorial/main/models/"
pretrained_model_path = "pfn_pretrained.pth"     # weights trained beforehand (downloaded)
pretrained_history_path = "pfn_history.json"     # their training/validation losses (downloaded)
model_path = "pfn_trained.pth"                   # where we save our own model when training from scratch

torch.manual_seed(0)  # same initial weights every time
model = PseudoParticleFlow().to(device)
model
```

We define the optimizer that we'll use for training, and the training settings: the maximum number of epochs and how many epochs without improvement on the validation loss we tolerate before lowering the learning rate (and, after twice as many, stopping).

```python
learning_rate = 0.001
optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)

nepochs = 30
npatience = 5
```

A quick check of how the mask works: the first line counts the real particles in the first two jets, the second averages their (standardised) four-momenta over the real particles only, ignoring the zero-padding. This is exactly the pooling step in the network's `forward`.

```python
print(torch.einsum('ik,ik->i',train_torch[:2][2],train_torch[:2][2]))
print(torch.einsum('ijk,ij->ik',train_torch[:2][0],train_torch[:2][2])/(torch.einsum('ik,ik->i',train_torch[:2][2],train_torch[:2][2]).unsqueeze(1)))
```

And let's train (or load). We use the validation dataset to monitor overfitting: after every epoch we compute the validation loss, keep the model with the lowest one, lower the learning rate when it stops improving, and stop early when it doesn't improve for a while.

Note that this means we are *choosing* the model with the validation dataset, so its validation performance is slightly optimistic. For an unbiased final number, evaluate on the test dataset, which we haven't touched.

```python
import json

if not TRAIN_FROM_SCRATCH:
    try:
        for path in [pretrained_model_path, pretrained_history_path]:
            if not os.path.isfile(path):
                urllib.request.urlretrieve(PRETRAINED_URL + path, path)
    except Exception as error:
        print(f"Could not download the pre-trained model ({error}); training from scratch instead")
        TRAIN_FROM_SCRATCH = True

if TRAIN_FROM_SCRATCH:
    history = {"train_loss": [], "val_loss": [], "learning_rate": []}
    best_val_loss = np.inf
    epochs_without_improvement = 0
    for epoch in range(1, nepochs+1):
        loss = train(model, optimizer,train_torch,batch_size=1024)
        val_loss = test(model,val_torch,batch_size=1024)
        history["train_loss"].append(loss)
        history["val_loss"].append(val_loss)
        history["learning_rate"].append(learning_rate)
        print(f'Epoch: {epoch:02d}, Loss: {loss:.6f}, Val Loss: {val_loss:.6f}, Learning Rate: {learning_rate:.6f}')

        if val_loss < best_val_loss:
            best_val_loss = val_loss
            epochs_without_improvement = 0
            torch.save(model.state_dict(),model_path)  # keep the best model so far
        else:
            epochs_without_improvement += 1

        if epochs_without_improvement == npatience:
            learning_rate *= 0.1
            for group in optimizer.param_groups:
                group['lr'] = learning_rate
        if epochs_without_improvement == 2*npatience:
            print('Early stopping')
            break

    # go back to the best model, and keep the history next to it
    model.load_state_dict(torch.load(model_path, map_location=device))
    with open(model_path.replace(".pth", "_history.json"), "w") as f:
        json.dump(history, f)
else:
    model.load_state_dict(torch.load(pretrained_model_path, map_location=device))
    with open(pretrained_history_path) as f:
        history = json.load(f)
    print("Loaded the pre-trained model")
```

### Learning curves

The training history tells us a lot about how training went. Things to look for:
* Both losses should go down and flatten out. If the validation loss starts going *up* while the training loss keeps going down, the network is **overfitting**: memorising the training jets instead of learning features that generalise.
* A big gap between the two curves is another sign of overfitting.
* Sudden drops often coincide with the learning rate being lowered.

```python
epochs = np.arange(1, len(history["train_loss"])+1)
best_epoch = epochs[np.argmin(history["val_loss"])]

fig, ax = plt.subplots()
ax.plot(epochs, history["train_loss"], 'o-', label='Training')
ax.plot(epochs, history["val_loss"], 'o-', label='Validation')
ax.axvline(best_epoch, color='gray', linestyle=':', label=f'Best model (epoch {best_epoch})')
for epoch, (lr_before, lr_after) in enumerate(zip(history["learning_rate"][:-1], history["learning_rate"][1:]), start=2):
    if lr_after < lr_before:
        ax.axvline(epoch, color='black', linestyle='--', alpha=0.3)
ax.set_xlabel('Epoch')
ax.xaxis.set_major_locator(plt.MaxNLocator(integer=True))
ax.set_ylabel('Binary cross-entropy loss')
ax.set_title('Learning curves (dashed lines: learning rate lowered)')
ax.legend()
plt.savefig('learning_curves.png')
plt.show()
```

```python
def plot_latent_hist(nlatent_dim, ax=None):
    ax = ax or plt.gca()
    values = val_latent_variables_NN[:,nlatent_dim]
    bins = np.linspace(*np.percentile(values,[0.5,99.5]),50)
    ax.hist(values[val_labels_NN==0],bins=bins,histtype='step',color='blue',label='QCD',density=True)
    ax.hist(values[val_labels_NN==1],bins=bins,histtype='step',color='red',label='Top',density=True)
    ax.set_xlabel('Latent variable '+str(nlatent_dim+1))

plot_latent_hist(0)
plt.legend(loc='upper right')
plt.ylabel('Density')
plt.savefig('latent_1.png')
plt.show()
```

```python
plot_latent_hist(1)
plt.legend(loc='upper right')
plt.ylabel('Density')
plt.savefig('latent_2.png')
plt.show()
```
